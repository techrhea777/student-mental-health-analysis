```python
import pandas as pd


def load_data(path: str) -> pd.DataFrame:
    df = pd.read_csv(path)

    # Standardize column names for easier plotting
    df = df.rename(columns={
        "avg_socialconnectedness": "avg_social"
    })

    # Basic validation
    required_columns = {
        "stay",
        "student_type",
        "avg_depression",
        "avg_social",
        "avg_stress",
    }

    missing = required_columns - set(df.columns)
    if missing:
        raise ValueError(f"Missing required columns: {sorted(missing)}")

    return df
```

```python
import os
import matplotlib.pyplot as plt
import pandas as pd


OUTPUT_DIR = "output"
os.makedirs(OUTPUT_DIR, exist_ok=True)


def plot_depression_trend(df: pd.DataFrame) -> None:
    plt.figure(figsize=(8, 5))

    for student_type in df["student_type"].unique():
        subset = df[df["student_type"] == student_type]
        plt.plot(subset["stay"], subset["avg_depression"], label=student_type)

    plt.xlabel("Length of Stay")
    plt.ylabel("Average Depression")
    plt.title("Depression Trends by Student Type")
    plt.legend()
    plt.tight_layout()
    plt.savefig(f"{OUTPUT_DIR}/depression_trend.png")
    plt.show()
    plt.close()


def plot_stress_trend(df: pd.DataFrame) -> None:
    plt.figure(figsize=(8, 5))

    for student_type in df["student_type"].unique():
        subset = df[df["student_type"] == student_type]
        plt.plot(subset["stay"], subset["avg_stress"], label=student_type)

    plt.xlabel("Length of Stay")
    plt.ylabel("Average Stress")
    plt.title("Stress Trends by Student Type")
    plt.legend()
    plt.tight_layout()
    plt.savefig(f"{OUTPUT_DIR}/stress_trend.png")
    plt.show()
    plt.close()


def plot_social_trend(df: pd.DataFrame) -> None:
    plt.figure(figsize=(8, 5))

    for student_type in df["student_type"].unique():
        subset = df[df["student_type"] == student_type]
        plt.plot(subset["stay"], subset["avg_social"], label=student_type)

    plt.xlabel("Length of Stay")
    plt.ylabel("Average Social Connectedness")
    plt.title("Social Connectedness Trends by Student Type")
    plt.legend()
    plt.tight_layout()
    plt.savefig(f"{OUTPUT_DIR}/social_trend.png")
    plt.show()
    plt.close()


def plot_comparison_bar(df: pd.DataFrame) -> None:
    avg_values = df.groupby("student_type")[[
        "avg_depression",
        "avg_social",
        "avg_stress"
    ]].mean()

    ax = avg_values.plot(kind="bar", figsize=(8, 5))
    ax.set_title("Overall Mental Health Comparison")
    ax.set_ylabel("Average Score")
    ax.set_xlabel("Student Type")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.savefig(f"{OUTPUT_DIR}/comparison_bar.png")
    plt.show()
    plt.close()


def plot_multi_metric(df: pd.DataFrame) -> None:
    plt.figure(figsize=(8, 5))

    subset = df[df["student_type"] == "International"]

    plt.plot(subset["stay"], subset["avg_depression"], label="Depression")
    plt.plot(subset["stay"], subset["avg_social"], label="Social Connectedness")
    plt.plot(subset["stay"], subset["avg_stress"], label="Stress")

    plt.xlabel("Length of Stay")
    plt.ylabel("Average Score")
    plt.title("Mental Health Metrics for International Students")
    plt.legend()
    plt.tight_layout()
    plt.savefig(f"{OUTPUT_DIR}/multi_metric.png")
    plt.show()
    plt.close()
```
